# A. Setting up for Scrapping Taks

```python
'''STEP 1: Load Necessary libraries'''
import requests
from bs4 import BeautifulSoup
```

```python
'''STEP 2: Define the URL of the Y Combinator jobs page'''
url = "https://www.ycombinator.com/blog/jobs"
```

```python
'''STEP 3: Developing User Agent to Mimic a Browser'''

# Set a User-Agent to look like a normal browser
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                  "AppleWebKit/537.36 (KHTML, like Gecko) "
                  "Chrome/120.0 Safari/537.36"
}
```

```python
''' STEP 4: Knowing whether the page is fetched successfully or not'''

# Fetch the page
response = requests.get(url, headers=headers)

# Check the status
if response.status_code == 200:
    print("Page fetched successfully!")
else:
    print(f"Failed to fetch page. Status code: {response.status_code}")
```

```python
'''STEP 5": Parse and Display the HTML Content'''
# Parse HTML
soup = BeautifulSoup(response.text, "html.parser")
print(soup)

# Optional: print the first 500 characters to see the structure
print(soup.prettify()[:500])
```

# STEP B -->

```python
# Find all job/blog entries
entries = soup.find_all("div", class_="mb-6 mr-6 flex flex-1 flex-col justify-between")

data = []

for entry in entries:
    # Title
    title_tag = entry.find("p", class_="text-3xl font-semibold text-gray-900")
    title = title_tag.text.strip() if title_tag else None

    # Author
    author_tag = entry.find("a", class_="text-blue-500 hover:underline")
    author = author_tag.text.strip() if author_tag else None

    # Date
    date_tag = entry.find("span", class_="ml-4 text-sm font-normal")
    date = date_tag.text.strip() if date_tag else None

    # Summary
    summary_tag = entry.find("p", class_="mt-3 text-base text-gray-800")
    summary = summary_tag.text.strip() if summary_tag else None

    # Read more link (full URL)
    link_tag = entry.find("a", class_="inline-flex text-sm font-medium text-blue-500 md:text-base")
    link = "https://www.ycombinator.com" + link_tag['href'] if link_tag else None

    # Append to list
    data.append({
        "title": title,
        "author": author,
        "date": date,
        "summary": summary,
        "link": link
    })

```

```python


# Quick check
print(data)  # print first 2 entries
```

```python
import pandas as pd

# Convert list of dicts into a DataFrame
df = pd.DataFrame(data)

# Optional: reorder columns
df = df[["title", "author", "date", "summary", "link"]]

# Quick preview

```

# MOST Important handling pagination

```python
import requests
from bs4 import BeautifulSoup

BASE_URL = "https://www.ycombinator.com/blog/jobs"

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                  "AppleWebKit/537.36 (KHTML, like Gecko) "
                  "Chrome/120.0 Safari/537.36"
}

def fetch_page(page):
    if page == 1:
        url = BASE_URL
    else:
        url = f"{BASE_URL}?page={page}"
    
    print(f"Fetching: {url}")
    response = requests.get(url, headers=headers)
    
    if response.status_code != 200:
        print("Error:", response.status_code)
        return None
    
    return BeautifulSoup(response.text, "html.parser")
```

```python
def extract_entries(soup):
    entries = soup.find_all("div", class_="mb-6 mr-6 flex flex-1 flex-col justify-between")
    
    page_data = []

    for entry in entries:
        title_tag = entry.find("p", class_="text-3xl font-semibold text-gray-900")
        author_tag = entry.find("a", class_="text-blue-500 hover:underline")
        date_tag = entry.find("span", class_="ml-4 text-sm font-normal")
        summary_tag = entry.find("p", class_="mt-3 text-base text-gray-800")
        link_tag = entry.find("a", class_="inline-flex text-sm font-medium text-blue-500 md:text-base")

        page_data.append({
            "title": title_tag.text.strip() if title_tag else None,
            "author": author_tag.text.strip() if author_tag else None,
            "date": date_tag.text.strip() if date_tag else None,
            "summary": summary_tag.text.strip() if summary_tag else None,
            "link": "https://www.ycombinator.com" + link_tag["href"] if link_tag else None
        })

    return page_data
```

```python
import time

all_entries = []
page = 1

while True:
    soup = fetch_page(page)
    if soup is None:
        break

    page_data = extract_entries(soup)

    # Stop when page is empty
    if not page_data:
        print("No more pages. Stopping.")
        break

    all_entries.extend(page_data)

    print(f"Extracted {len(page_data)} entries from page {page}")
    
    page += 1
    time.sleep(2)  # polite delay
```

```python
import pandas as pd

df = pd.DataFrame(all_entries)
df = df[["title", "author", "date", "summary", "link"]]

df
```

# Advanced and FUll SCRAPPER

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
import random

BASE_URL = "https://www.ycombinator.com/blog"
HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/117.0.0.0 Safari/537.36",
    "Accept-Language": "en-US,en;q=0.9"
}

def scrape_blog_page(page_number):
    url = f"{BASE_URL}?page={page_number}"
    print(f"Fetching {url}")
    response = requests.get(url, headers=HEADERS)
    
    if response.status_code != 200:
        print(f"Failed to fetch page {page_number}, status code: {response.status_code}")
        return None

    soup = BeautifulSoup(response.text, "html.parser")

    # Try page 1 structure
    posts = soup.find_all("div", class_="flex flex-col overflow-hidden rounded shadow")
    # If none found, try page 2+ structure
    if not posts:
        posts = soup.find_all("div", class_="mb-6 mr-6 flex flex-1 flex-col justify-between")
    
    if not posts:
        return None  # No posts found

    data = []
    for post in posts:
        # Try headline
        headline_tag = post.find("p", class_="text-xl font-semibold text-gray-900")
        if not headline_tag:
            headline_tag = post.find("p", class_="text-3xl font-semibold text-gray-900")
        headline = headline_tag.get_text(strip=True) if headline_tag else "N/A"

        # Try description
        description_tag = post.find("p", class_="mt-3 text-base text-gray-700")
        if not description_tag:
            description_tag = post.find("p", class_="mt-3 text-base text-gray-800")
        description = description_tag.get_text(strip=True) if description_tag else "N/A"

        # Try author
        author_tag = post.find("a", class_="hover:underline")
        if not author_tag:
            author_tag = post.find("a", class_="text-blue-500 hover:underline")
        author = author_tag.get_text(strip=True) if author_tag else "N/A"

        # Try date
        date_tag = post.find("span")
        date = date_tag.get_text(strip=True) if date_tag else "N/A"

        data.append({
            "Headline": headline,
            "Description": description,
            "Author": author,
            "Date": date
        })

    return data

def scrape_all_pages(min_sleep=2, max_sleep=5):
    all_posts = []
    page_number = 1

    while True:
        posts = scrape_blog_page(page_number)
        if posts is None:
            print("No more posts found. Ending scraping.")
            break

        all_posts.extend(posts)

        sleep_time = random.uniform(min_sleep, max_sleep)
        print(f"Sleeping for {sleep_time:.2f} seconds...")
        time.sleep(sleep_time)

        page_number += 1

    return all_posts

# Scrape all pages safely
all_posts = scrape_all_pages(min_sleep=2, max_sleep=2)

```

```python
df = pd.DataFrame(all_posts)
#df.head()
df.tail()
```

```python

df.to_csv("yc_blog_all_posts.csv", index=False)
print(f"Scraping complete! {len(all_posts)} posts saved to yc_blog_all_posts.csv")
```
