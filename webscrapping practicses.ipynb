```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# Send a GET request to the website
url = 'https://quotes.toscrape.com/'
response = requests.get(url)

# Parse the response text with BeautifulSoup
soup = BeautifulSoup(response.text, 'html.parser')

# Find all quotes on the page
quotes = soup.find_all('div', class_='quote')

# List to store scraped data
data = []

# Loop through each quote and extract the details
for quote in quotes:
    text = quote.find('span', class_='text').text
    author = quote.find('small', class_='author').text
    tags = [tag.text for tag in quote.find_all('a', class_='tag')]
    
    # Append the data
    data.append({'Quote': text, 'Author': author, 'Tags': ', '.join(tags)})

# Convert the data into a DataFrame
df = pd.DataFrame(data)

# Save the DataFrame to a CSV file
# df.to_csv('quotes_first_page.csv', index=False)

print("Data saved to 'quotes_first_page.csv'")
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# List to store all scraped data
all_data = []

# Loop to handle pagination
page_number = 1

while True:
    url = f'https://quotes.toscrape.com/page/{page_number}/'
    response = requests.get(url)
    
    # If the response status is 200 (successful), proceed
    if response.status_code == 200:
        # Parse the response text
        soup = BeautifulSoup(response.text, 'html.parser')
        
        # Find all quotes on the page
        quotes = soup.find_all('div', class_='quote')
        
        # Extract data from each quote
        for quote in quotes:
            text = quote.find('span', class_='text').text
            author = quote.find('small', class_='author').text
            tags = [tag.text for tag in quote.find_all('a', class_='tag')]
            
            # Append the data
            all_data.append({'Quote': text, 'Author': author, 'Tags': ', '.join(tags)})
        
        # Check if there is a next page
        next_page = soup.find('li', class_='next')
        if next_page:
            page_number += 1  # Move to the next page
        else:
            break  # No more pages, break the loop
    else:
        print("Failed to retrieve page", page_number)
        break

# Convert the data into a DataFrame
df_all = pd.DataFrame(all_data)

# Save the DataFrame to a CSV file
df_all.to_csv('quotes_all_pages.csv', index=False)

print("Data saved to 'quotes_all_pages.csv'")
```
